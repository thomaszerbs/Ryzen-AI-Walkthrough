### A1 — Load Data, Model & Export to ONNX
`prepare_model_data.py` automates three steps:
1. Download CIFAR-10 dataset
2. Load pre-trained ResNet50 (ImageNet weights), adapt the FC head for CIFAR-10 (10 classes)
3. Export to ONNX — **opset 17 required** by VitisAI EP / VAIML

> **Why ResNet50?** Lightweight CIFAR-10 dataset (~60 s compile time) vs. large COCO dataset (15+ min for YOLOv8). Ideal for switching execution providers live.

```python
# Key APIs from prepare_model_data.py

import torch
import torch.nn as nn
from torchvision.models import resnet50, ResNet50_Weights

# --- 1. Load pre-trained ResNet50 (ImageNet weights) ---
def load_resnet_model():
    weights = ResNet50_Weights.DEFAULT
    resnet = resnet50(weights=weights)
    # Replace final FC: ImageNet has 1000 classes, CIFAR-10 has 10
    # Keep full backbone (conv layers), swap only the head: 2048 → 64 → 10
    resnet.fc = torch.nn.Sequential(
        torch.nn.Linear(2048, 64),
        torch.nn.ReLU(inplace=True),
        torch.nn.Linear(64, 10)
    )
    return resnet

# --- 2. Export to ONNX ---
def export_to_onnx(model, models_dir):
    model.to('cpu')
    dummy_inputs = torch.randn(1, 3, 32, 32)    # CIFAR-10 input: 3-channel, 32x32
    tmp_model_path = str(models_dir / 'resnet_trained_for_cifar10.onnx')
    torch.onnx.export(
        model,
        dummy_inputs,
        tmp_model_path,
        export_params=True,
        opset_version=17,                       # VitisAI EP / VAIML requires opset 17
        input_names=['input'],
        output_names=['output'],
        dynamic_axes={'input': {0: 'batch_size'}, 'output': {0: 'batch_size'}}
    )
    print(f'Exported → {tmp_model_path}')

# Show model FC head (pre-trained weights download takes a moment)
model = load_resnet_model()
print(f'FC head: {model.fc}')
```

```python
# Run prepare_model_data.py
# Downloads CIFAR-10, loads trained checkpoint, and exports to ONNX
# Output: models/resnet_trained_for_cifar10.onnx (FP32, ready for Quark or VAIML)
!python prepare_model_data.py
```

### A2 — Quantize with AMD Quark (INT8)
`resnet_quantize.py` — post-training quantization (PTQ), Quark converts FP32 ONNX → INT8 ONNX

| Config | Description |
|--------|-------------|
| `XINT8` | Symmetric INT8, power-of-two scales — **fastest INT8 mode** |
| `A8W8` | Asymmetric activations, symmetric weights |
| `A16W8` | 16-bit activations, 8-bit weights |

> **Calibration matters**: Without calibration (random_quantize), L2 loss ≈ 30. With calibration: ≈ 9.78. With AdaRound: ≈ 1.43.

```python
# Key APIs from resnet_quantize.py

from quark.onnx.quantization.config import Config, get_default_config
from quark.onnx import ModelQuantizer
from resnet_quantize import resnet_calibration_reader

# 1. Pick quantization config
#    XINT8 = symmetric INT8, power-of-two scales — fastest INT8 mode
quant_config = get_default_config('XINT8')
config = Config(global_quant_config=quant_config)
print(f'Quant config: {config}')

# 2. Create ModelQuantizer
quantizer = ModelQuantizer(config)

# 3. Run PTQ with calibration data
#    CalibrationDataReader feeds representative images to calibrate activation ranges
#    dr = resnet_calibration_reader('data/')
#    quantizer.quantize_model(
#        'models/resnet_trained_for_cifar10.onnx',   # input: FP32 ONNX
#        'models/resnet_quantized.onnx',             # output: INT8 ONNX
#        dr                                          # calibration data reader
#    )
#    → run via subprocess below (~1-2 minutes)
```

### A3 — Inference: CPU Baseline → NPU
`predict.py` creates an ONNX Runtime InferenceSession and runs predictions on 10 CIFAR-10 test images.

**Best practice**: run CPU first to verify the model, then switch to NPU.
- **CPU** (`CPUExecutionProvider`): no driver needed, confirms model logic is correct
- **NPU** (`VitisAIExecutionProvider`): compiles INT8 model on first run (~60 s), cached to `./modelcachekey/`

Provider options for **INT8 on STX/KRK**:

```python
# Key APIs from predict.py

import onnxruntime as ort
import numpy as np
from pathlib import Path

model_path = 'models/resnet_quantized.onnx'
cache_dir  = Path('.').resolve()

# --- CPU (default) ---
providers        = ['CPUExecutionProvider']
provider_options = [{}]

# --- NPU via Vitis AI EP (STX/KRK) ---
# providers = ['VitisAIExecutionProvider']
# provider_options = [{
#     'cache_dir': str(cache_dir),      # compiled NPU binary saved here
#     'cache_key': 'modelcachekey',     # change to force recompile when model changes
#     'enable_cache_file_io_in_mem': '0',
# }]

# Create ONNX RT session
session_options = ort.SessionOptions()
session_options.log_severity_level = 1  # 0=Verbose 1=Info 2=Warning 3=Error 4=Fatal

session = ort.InferenceSession(model_path,
                               sess_opts=session_options,
                               providers=providers,
                               provider_options=provider_options)

print(f'Active provider: {session.get_providers()}')

# Run inference — identical API regardless of CPU or NPU
# outputs = session.run(None, {'input': input_data})
# predicted_class = np.argmax(outputs[0])
```

### C2 — NPU Management with `xrt-smi`
`xrt-smi` is the NPU management utility — analogous to `nvidia-smi` for NVIDIA GPUs.

| Command | Purpose |
|---------|----------|
| `xrt-smi examine` | System info: OS, XRT version, NPU driver/firmware, device BDF and name |
| `xrt-smi examine --report platform` | Performance mode and estimated power draw (Watts) |
| `xrt-smi examine --report aie-partitions` | Run **while a model is active** — NPU partition and column occupancy |
| `xrt-smi validate --run all` | Standalone sanity tests: no-op latency, DMA throughput, gemm INT8 TOPS |
| `xrt-smi configure --pmode performance` | Set performance mode before benchmarking |

### C3 — Benchmarking
`benchmark()` in `run_inference.py` is model-agnostic — it only needs an ORT session, input name, and input tensor.  
Works identically for ResNet50 INT8, ResNet50 BF16, and YOLOv8m by swapping session and model path.

**Pattern**: 10 warmup runs (driver/JIT settling) → 100 timed runs → report avg latency + FPS.

```python
# Key code from run_inference.py — benchmark() function

import time

def benchmark(session, input_name, input_img, num_inference=100):
    """Model-agnostic benchmark: 10 warmup + N timed inference runs."""

    # Warmup — discard first runs (driver/JIT settling)
    for _ in range(10):
        session.run(None, {input_name: input_img})

    # Timed inference loop
    time_list = []
    for _ in range(num_inference):
        start = time.time()
        output = session.run(None, {input_name: input_img})
        time_list.append(time.time() - start)

    avg_time_per_inference = sum(time_list) / num_inference
    print(f'Avg time for each inference run: {avg_time_per_inference:.3f} seconds')
    print(f'Model performance: {1/avg_time_per_inference:.1f} FPS')

# Usage: benchmark(ort_session, input_name, input_img)
# Swap session+model path to benchmark any model on any EP
print('benchmark() defined — call after creating your ORT session')
```

```python
# Benchmark YOLOv8m BF16 on NPU: 10 warmup + 100 timed runs
# Expected: ~0.075 s latency / 13.4 FPS (vs 0.486 s / 2.1 FPS on CPU)
!xrt-smi configure --pmode performance
!python run_inference.py \
    --model_input models/yolov8m_BF16.onnx \
    --input_image test_image.jpg \
    --output_image test_output.jpg \
    --device npu-bf16 \
    --benchmark
```
