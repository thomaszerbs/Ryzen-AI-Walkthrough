**Key APIs** (`prepare_model_data.py`):

```python
from torchvision.models import resnet50, ResNet50_Weights

# 1. Load ResNet50 with pre-trained ImageNet weights
weights = ResNet50_Weights.DEFAULT
resnet = resnet50(weights=weights)

# Replace FC head: ImageNet has 1,000 classes, CIFAR-10 has 10
# Keep the backbone, swap head: 2048 → 64 → 10
resnet.fc = nn.Sequential(nn.Linear(2048, 64), nn.ReLU(inplace=True), nn.Linear(64, 10))
```

```python
# 2. Export to ONNX function
def export_to_onnx(model, models_dir):
    model.to("cpu")
    dummy_inputs = torch.randn(1, 3, 32, 32)    # CIFAR-10 input size
    input_names = ['input']
    output_names = ['output']
    dynamic_axes = {'input': {0: 'batch_size'}, 'output': {0: 'batch_size'}}
    tmp_model_path = str(models_dir / "resnet_trained_for_cifar10.onnx")
    torch.onnx.export(
        model,
        dummy_inputs,
        tmp_model_path,
        export_params=True,
        opset_version=17,              # VitisAI EP / VAIML requires opset 17
        input_names=input_names,
        output_names=output_names,
        dynamic_axes=dynamic_axes,
    )
```

> **ONNX opset 17** is required by VitisAI EP / VAIML. If your model uses a different opset version, convert it using the [ONNX Version Converter](https://github.com/onnx/onnx/blob/main/docs/VersionConverter.md).

**Output**: `models/resnet_trained_for_cifar10.onnx` — FP32 ONNX, ready for Quark quantization (INT8) or VAIML compilation (BF16).

### A3 — Inference: CPU Baseline → NPU

`predict.py` creates an ONNX Runtime `InferenceSession` and runs predictions on 10 CIFAR-10 test images.  
**Best practice**: run CPU first to verify the model is correct, then switch to NPU.

**Key APIs** (`predict.py`):

```python
import onnxruntime as ort

# --- CPU (default, no flag) ---
providers        = ['CPUExecutionProvider']
provider_options = [{}]

# --- NPU via Vitis AI EP (--ep npu) ---
providers = ['VitisAIExecutionProvider']
provider_options = [{
    'cache_dir': str(cache_dir),       # compiled NPU binary saved here
    'cache_key': 'modelcachekey',     
    'enable_cache_file_io_in_mem': '0',
}]

# Create ONNX RT session
session_options = ort.SessionOptions()
session_options.log_severity_level = 1  # 0=Verbose 1=Info 2=Warning 3=Error 4=Fatal

session = ort.InferenceSession(
    model.SerializeToString(),
    sess_options=session_options,
    providers=providers,
    provider_options=provider_options
)

# Run inference — identical API regardless of CPU or NPU
outputs = session.run(None, {'input': input_data})
predicted_class = np.argmax(outputs[0])
```

**Provider options for INT8 on STX/KRK** (focused):
- `cache_dir` & `cache_key` — compiled binary cache location and unique model key
- `opt_level` — compiler optimization level (0, 1, 2, 3, 65536), INT8 only
- `enable_cache_file_io_in_mem` — keep compiled model in memory (1) or save to disk (0)

#### Full Vitis AI EP Provider Options

| Option | Description |
|--------|-------------|
| `config_file` | Path to compilation config JSON (required for BF16 compilation) |
| `target` | Compiler backend — `X2` (STX/KRK, default) or `X1` (PHX/HPT legacy integer) |
| `xclbin` | xclbin path — required for INT8 on PHX/HPT |
| `cache_dir` | Directory where compiled NPU artifacts are stored |
| `cache_key` | Subfolder name within `cache_dir` for this model's compiled artifacts |
| `enable_cache_file_io_in_mem` | `1` = keep compiled model in memory, `0` = save to disk |
| `opt_level` | Compiler optimization level (0, 1, 2, 3, 65536) — INT8 only |
| `encryption_key` | 256-bit key for encrypting the EP context model |
| `ai_analyzer_visualization` | Enable compilation-time analysis data collection |
| `ai_analyzer_profiling` | Enable inference-time analysis data collection |

Source: [Model Compilation and Deployment — Ryzen AI Software 1.7.1](https://ryzenai.docs.amd.com/en/latest/modelrun.html)

> **First NPU run** compiles the INT8 model (~60 s) and caches to `./modelcachekey/`. Subsequent runs load from cache in under 1 second. Predictions should match CPU output.

#### Improve PTQ Accuracy

After post-training quantization (PTQ), several techniques can help recover accuracy:

| Technique | Description |
|----------|------------|
| Cross-Layer Equalization (CLE) | Balances weights across adjacent layers to reduce quantization sensitivity. |
| AdaRound | Learns optimized rounding of weights to better approximate FP32 behavior. |
| AdaQuant | Applies lightweight fine-tuning to minimize the gap between quantized and original model outputs. |
| SmoothQuant | Redistributes scale between activations and weights to stabilize quantization. |

**L2 loss vs FP32 (ResNet50 A8W8):**

| | No calibration | + Calibration | + AdaRound | + AdaQuant |
|---|---|---|---|---|
| **L2 loss** | 30.26 | 9.78 | 1.43 | **1.15** |

> **Reference**: [Accuracy Improvement Algorithms — AMD Quark 0.11.1 documentation](https://quark.docs.amd.com/latest/onnx/accuracy_improvement_algorithms.html)

### D2 — Benchmarking

Two quick/lightweight benchmarking options to estimate performance:
1. `benchmark()` ONNX Runtime wrapper
2. `onnx-benchmark` tool in RyzenAI-SW repo

#### 1. `benchmark()` ONNX Runtime wrapper 
- Built in `--benchmark` flag in `run_inference.py`.
- Lightweight function that calls ONNX inference session.
- Measures end-to-end app performance (pre/post-processing included).
- Uses real input data.

```python
def benchmark(session, input_name, input_img, num_inference=100):
    # Warmup — 10 runs (overhead: kernel intialization, caching, etc.)
    for _ in range(10):
        session.run(None, {input_name: input_img})

    # Timed inference loop
    time_list = []
    for _ in range(num_inference):
        start = time.time   # capture current timestamp
        output = session.run(None, {input_name: input_img})   # run inference
        time_list.append(time.time() - start)   # compujte elapsed time and store

    avg = sum(time_list) / num_inference
    print(f'Avg time: {avg:.3f} s  |  {1/avg:.1f} FPS')
```

```bash
    --model_path e.g., models/resnet_quantized.onnx \
    --execution_provider e.g, VitisAIEP \
    --config $VAIP_CONFIG \
    --autoquant 0 --renew 0 --num 100 --timelimit 10
```
| Option | Description |
|--------|-------------|
| `model_path` | Path to the ONNX model to benchmark |
| `execution_provider` | Hardware target (e.g., `VitisAIEP` = NPU, `CPU` = baseline) |
| `config` | Vitis AI config file for NPU compilation/runtime |
| `autoquant 0` | Disable automatic quantization (use model as-is) |
| `renew 0` | Reuse compiled model cache (avoid recompilation overhead) |
| `num 100` | Number of input samples to process |
| `timelimit 10` | Maximum runtime (seconds) for the benchmark |

### Performance Summary

````python
import json, os
import pandas as pd

BENCH = '/scratch/thozerbs/git/RyzenAI-SW/onnx-benchmark'

files = {
    ('ResNet50 INT8', 'NPU'): 'resnet_int8_npu.json',
    ('ResNet50 INT8', 'CPU'): 'resnet_int8_cpu.json',
    ('ResNet50 BF16', 'NPU'): 'resnet_bf16_npu.json',
    ('ResNet50 BF16', 'CPU'): 'resnet_bf16_cpu.json',
    ('YOLOv8m BF16',  'NPU'): 'yolo_bf16_npu.json',
    ('YOLOv8m BF16',  'CPU'): 'yolo_bf16_cpu.json',
}

rows = []
for (model, device), fname in files.items():
    path = os.path.join(BENCH, fname)
    if os.path.exists(path):
        d = json.load(open(path))
        perf = d['results']['performance']
        rows.append({
            'Model': model, 'Device': device,
            'Throughput (FPS)': round(perf['total_throughput'], 1),
            'Latency (ms)':     round(perf['average_latency'], 2),
        })
    else:
        rows.append({'Model': model, 'Device': device,
                     'Throughput (FPS)': '—', 'Latency (ms)': '—'})

df = pd.DataFrame(rows).set_index(['Model', 'Device'])
df
````

<!-- ##### BF16 = original FP32 model. INT8 = quantized model with extra conversion ops (QuantizeLinear and DequantizeLinear), CPU has no INT8 acceleration, runs slower. -->
