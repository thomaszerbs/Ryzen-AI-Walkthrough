# Part 1: ResNet50 INT8 - Quark Quantization + NPU Inference

This section demonstrates the **INT8 quantization** flow:
1. Download CIFAR-10 data and prepare a ResNet50 model (export to ONNX opset 17)
2. Quantize with AMD Quark (XINT8 configuration)
3. Run inference on CPU and NPU using ONNX Runtime with VitisAI Execution Provider

### Step 1: Prepare Model & Data

**`prepare_model_data.py`** does the following:
- Downloads the CIFAR-10 dataset
- Loads a pretrained ResNet50 with ImageNet weights, replaces the FC head for 10-class CIFAR-10
- Exports to ONNX with **opset 17** (required for VitisAI EP)

```python
# Key APIs:
from torchvision.models import ResNet50_Weights, resnet50

resnet = resnet50(weights=ResNet50_Weights.DEFAULT)
resnet.fc = torch.nn.Sequential(
    torch.nn.Linear(2048, 64),
    torch.nn.ReLU(inplace=True),
    torch.nn.Linear(64, 10)
)

# Export to ONNX - opset_version=17 is critical for VitisAI EP
torch.onnx.export(model, dummy_inputs, output_path,
                  export_params=True, opset_version=17,
                  input_names=['input'], output_names=['output'],
                  dynamic_axes={'input': {0: 'batch_size'}, 'output': {0: 'batch_size'}})
```

````python
%%bash
cd /scratch/thozerbs/git/RyzenAI-SW/CNN-examples/getting_started_resnet/int8
python -u prepare_model_data.py
````

### Step 2: Quantize with AMD Quark

**`resnet_quantize.py`** performs post-training quantization (PTQ) using AMD Quark:

```python
from quark.onnx.quantization.config import Config, get_default_config
from quark.onnx import ModelQuantizer

# Get the XINT8 quantization configuration
quant_config = get_default_config("XINT8")
config = Config(global_quant_config=quant_config)

# Create quantizer and quantize
quantizer = ModelQuantizer(config)
quantizer.quantize_model(input_model_path, output_model_path, calibration_data_reader)
```

The calibration data reader feeds a subset of CIFAR-10 training images through the model to determine optimal quantization scale factors.

````python
%%bash
cd /scratch/thozerbs/git/RyzenAI-SW/CNN-examples/getting_started_resnet/int8
python -u resnet_quantize.py
````

### Step 3b: INT8 Inference on NPU

Now run the same quantized model on the **NPU** using the VitisAI Execution Provider:

```python
providers = ['VitisAIExecutionProvider']
provider_options = [{
    'cache_dir': str(cache_dir),        # Directory for compiled model cache
    'cache_key': 'modelcachekey',        # Unique key for this model's cache
    'enable_cache_file_io_in_mem': '0',  # Disk-based caching
}]

session = ort.InferenceSession(model.SerializeToString(),
                               sess_options=session_options,
                               providers=providers,
                               provider_options=provider_options)
```

On first run, the VitisAI EP **compiles** the model for the NPU (takes longer). Subsequent runs use the cached compilation.

# Part 2: ResNet50 BF16 - VAIML Auto-Casting + AI Analyzer

This section demonstrates the **BF16** flow using the VAIML compiler:
1. Compile the FP32 ONNX model - VAIML automatically converts to BF16
2. Run inference on CPU and NPU
3. Use AI Analyzer for graph visualization and profiling

The key difference from INT8: **no manual quantization needed!** The VAIML compiler handles FP32 -> BF16 conversion automatically during compilation.

### BF16 Configuration: `vitisai_config.json`

The VAIML compiler is configured via a JSON file. The critical setting is `enable_f32_to_bf16_conversion`:

```json
{
    "passes": [
        {"name": "init", "plugin": "vaip-pass_init"},
        {
            "name": "vaiml_partition",
            "plugin": "vaip-pass_vaiml_partition",
            "vaiml_config": {
                "enable_f32_to_bf16_conversion": true
            }
        }
    ],
    "target": "VAIML"
}
```

This tells the VAIML compiler to automatically cast FP32 operations to BF16 for NPU acceleration.

### Step 1: Compile BF16 Model

**`compile.py`** creates an ORT InferenceSession with VitisAI EP + the config file. This triggers VAIML compilation:

```python
provider_options_dict = {
    "config_file": "vitisai_config.json",    # VAIML config with BF16 conversion
    "cache_dir":   cache_dir,
    "cache_key":   cache_key,
    "ai_analyzer_visualization": True,         # Generate graph partition visualization
    "ai_analyzer_profiling": True,              # Generate per-op profiling data
}

session = onnxruntime.InferenceSession(
    onnx_model,
    sess_options=session_options,
    providers=["VitisAIExecutionProvider"],
    provider_options=[provider_options_dict]
)
```

````python
%%bash
cd /scratch/thozerbs/git/RyzenAI-SW/CNN-examples/getting_started_resnet/bf16
python -u compile.py --model models/resnet_trained_for_cifar10.onnx
````

### Step 2a: BF16 Inference on CPU

Run the FP32 model on CPU as a baseline (no BF16 conversion on CPU path):

```python
# CPU path - standard CPUExecutionProvider, no config file needed

provider_options_dict = {}
```

````python
%%bash
cd /scratch/thozerbs/git/RyzenAI-SW/CNN-examples/getting_started_resnet/bf16
python -u predict.py --ep cpu
````

### Step 2b: BF16 Inference on NPU

Now run on the NPU using the compiled BF16 model:

```python
# NPU path - VitisAI EP with config file + AI Analyzer
providers = ['VitisAIExecutionProvider']
provider_options_dict = {
    "config_file": config_file,
    "cache_dir": cache_dir,
    "cache_key": cache_key,
    "ai_analyzer_visualization": True,
    "ai_analyzer_profiling": True,
}

# Enable ORT profiler for CPU-side timing

# ... after inference ...
session.end_profiling()  # Finalize AI Analyzer profiling data
```

## NPU Management: `xrt-smi`

`xrt-smi` is the NPU management CLI (similar to `nvidia-smi` for GPUs). Key commands:

| Command | Description |
|---------|-------------|
| `xrt-smi examine` | Show NPU device info, firmware, drivers |
| `xrt-smi validate` | Run diagnostic tests on the NPU |
| `xrt-smi configure --pmode <mode>` | Set power mode (performance vs efficiency) |

## Summary

### Key Takeaways

| Flow | Quantization | Compiler | Config |
|------|-------------|----------|--------|
| **INT8** | AMD Quark PTQ (`get_default_config("XINT8")`) | VitisAI EP auto-compiles | `cache_dir`, `cache_key` |
| **BF16** | None needed | VAIML auto-converts FP32->BF16 | `vitisai_config.json` with `enable_f32_to_bf16_conversion: true` |

### Common API Pattern
```python
import onnxruntime as ort

session = ort.InferenceSession(
    model_path,
    providers=['VitisAIExecutionProvider'],
    provider_options=[{
        'config_file': 'vitisai_config.json',   # BF16 only
        'cache_dir': cache_dir,
        'cache_key': cache_key,
    }]
)
outputs = session.run(None, {input_name: input_data})
```

### Tools
- **AMD Quark** - Post-training quantization (INT8)
- **VAIML Compiler** - Automatic FP32->BF16 conversion
- **AI Analyzer** - Graph visualization + profiling
- **xrt-smi** - NPU device management
- **onnx-benchmark** - Standardized performance measurement
